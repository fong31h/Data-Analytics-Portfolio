```python
expansions = ['Genetic Apex','Mythical Island','Space-Time Smackdown','Triumphant Light',
             'Shining Revelry','Celestial Guardians','Extradimensional Crisis',
             'Eevee Grove','Wisdom of Sea and Sky','Secluded Springs']
expansion_modifiers = [.1,.2,.3,.4,.5,.6,.7,.8,.9,1]
```

```python
for i in range(len(expansions)):
    with open('Data_Files/best_names.txt', 'r') as f:
        ex_list = f.readlines()
        ex_list = [x.strip() for x in ex_list]
    unnormalized_score = []
    for name in ex_list:
        step2 = sp.filter_top_finishes(dropdown_df, name)
        step3 = step2[step2['Expansion'] == expansions[i]]
        step4_getscore = sp.make_score(step3)
        unnormalized_score.append(round(step4_getscore*expansion_modifiers[i],2))
    ex_list[ex_list.index('A1-36')] = 'Charizard ex: A1'
    ex_list[ex_list.index('A2b-10')] = 'Charizard ex: A2b'
    ex_list[ex_list.index('A1-96')] = 'Pikachu ex: A1'
    ex_list[ex_list.index('A2b-22')] = 'Pikachu ex: A2b'
    scores = pd.DataFrame({'Pokémon':ex_list})
    scores[f'{expansions[i]}'] = unnormalized_score
```
