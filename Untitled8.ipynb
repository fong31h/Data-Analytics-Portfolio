```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import datetime
import plotly.express as px
```

```python
dropdown_df = pd.read_csv('Data_Files/dropdown_df.csv')
```

```python
df = dropdown_df[dropdown_df['Decklists'].str.contains('Decidueye ex')]
x = df['Expansion'].unique()
y = [df[(df['Expansion'] == expansion) & (df['Player_Total'] > 200)]['Placement'].sort_values()[:10].mean() for expansion in df['Expansion'].unique()]

# comment
```

```python
fig = px.line(x=x,y=y)
fig.show()
```
