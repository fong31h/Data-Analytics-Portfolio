### creating image dictionary

```python
with open('Data_Files/all_ex.txt', 'r') as f:
    names = [line.strip() for line in f]
```

```python
names[2] = 'Charizard ex: A1' # Crimson Zard
names[38] = 'Charizard ex: A2b' # Stokezard
names[8] = 'Pikachu ex: A1' # Circle Circuit Pikachu
names[40] = 'Pikachu ex: A2b' # 150 dmg Pikachu
names.sort()
```

```python
images = os.listdir('dashboard/www/ex_images')
if '.DS_Store' in images:
    images.remove('.DS_Store')
```

```python
for i in range(len(images)):
    images[i] = 'www/ex_images/' + images[i]
images.sort()
```

```python
print(len(names))
print(len(images))
```

```python
image_dict = {name:{"src":image, "width":"250px"} for name,image in zip(names, images)}
```

```python
with open('Data_Files/image_dict.json', 'w') as f:
    json.dump(image_dict, f)
```
